```python
import pandas as pd

df = pd.read_csv('AB_NYC_2019.csv')
df.head()
```

```python
# 1. Dataset mein kitni rows aur columns hain?
print("Shape (Rows, Columns):", df.shape)

# 2. Har column ka data type aur non-null count kya hai?
print("\n--- Dataset Info ---")
df.info()

# 3. Kis column mein kitni missing (khali) values hain?
print("\n--- Missing Values ---")
print(df.isnull().sum())

# 4. Kya dataset mein koi duplicate rows hain?
print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# 1. 'name' aur 'host_name' ki missing values ko 'Unknown' se bharna
df['name'] = df['name'].fillna('Unknown')
df['host_name'] = df['host_name'].fillna('Unknown')

# 2. 'reviews_per_month' mein missing values ko 0 se bharna (kyunki in listings par 0 reviews hain)
df['reviews_per_month'] = df['reviews_per_month'].fillna(0)

# 3. 'last_review' ko proper datetime format mein convert karna
df['last_review'] = pd.to_datetime(df['last_review'])

# 4. Dobara check karna ki ab kitni missing values bachi hain
print("--- Cleaning ke baad Missing Values ---")
print(df.isnull().sum())
```

```python
# 1. Text columns se extra spaces (leading/trailing) hatana
text_cols = ['name', 'host_name', 'neighbourhood_group', 'neighbourhood', 'room_type']
for col in text_cols:
    df[col] = df[col].astype(str).str.strip()

# 2. Check karna ki kitni listings ka price 0 ya usse kam hai
print("Price <= 0 wali rows:", (df['price'] <= 0).sum())

# 3. Price > 0 aur minimum_nights <= 365 wali valid rows rakhna (Outliers hatana)
df = df[(df['price'] > 0) & (df['minimum_nights'] <= 365)]

# 4. Final cleaned dataset ka shape check karna
print("Final Cleaned Shape (Rows, Columns):", df.shape)
```

```python
# Cleaned data ko nayi CSV file mein save karna
df.to_csv('AB_NYC_2019_cleaned.csv', index=False)
print("Cleaned dataset successfully saved!")
```
